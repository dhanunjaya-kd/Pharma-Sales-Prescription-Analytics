```python
import pandas as pd
import numpy as np
from faker import Faker
from datetime import datetime, timedelta
import random

fake = Faker()

print("Setup Successful")
```

```python
# Doctors Dataset

specializations = [
    'Cardiologist',
    'Diabetologist',
    'General Physician',
    'Orthopedic',
    'Neurologist',
    'Gynecologist',
    'Dermatologist'
]

cities = [
    'Hyderabad',
    'Bangalore',
    'Chennai',
    'Mumbai',
    'Pune',
    'Delhi'
]

states = {
    'Hyderabad':'Telangana',
    'Bangalore':'Karnataka',
    'Chennai':'Tamil Nadu',
    'Mumbai':'Maharashtra',
    'Pune':'Maharashtra',
    'Delhi':'Delhi'
}

doctors = []

for i in range(1, 501):

    city = random.choice(cities)

    doctors.append([
        i,
        fake.name(),
        random.choice(specializations),
        fake.company() + " Hospital",
        city,
        states[city]
    ])

doctors_df = pd.DataFrame(
    doctors,
    columns=[
        'doctor_id',
        'doctor_name',
        'specialization',
        'hospital',
        'city',
        'state'
    ]
)

doctors_df.head()
```

```python
doctors_df.to_csv(
    '../01_Dataset/raw_data/doctors.csv',
    index=False
)

print("Doctors CSV Saved Successfully")
```

```python
# Products Dataset

categories = {
    'Cardiac': [
        'CardioPlus',
        'HeartSafe',
        'CardioCare',
        'PulseMax',
        'ArteryFlow',
        'CardioLife',
        'HeartWell',
        'CardioX',
        'VitaHeart',
        'CardioPrime'
    ],

    'Diabetes': [
        'DiaCare',
        'SugarFree',
        'GlucoMax',
        'DiaLife',
        'InsuCare',
        'GlucoSafe',
        'DiaPlus',
        'SugarShield',
        'DiaPrime',
        'GlucoCare'
    ],

    'Neuro': [
        'NeuroMax',
        'BrainPlus',
        'NeuroCare',
        'MindWell',
        'NeuroLife',
        'BrainSafe',
        'NeuroPrime',
        'NeuroX',
        'MindCare',
        'NeuroFlow'
    ],

    'Ortho': [
        'OrthoFlex',
        'BoneCare',
        'JointMax',
        'OrthoLife',
        'FlexiBone',
        'OrthoPrime',
        'JointCare',
        'BoneShield',
        'OrthoX',
        'FlexiCare'
    ],

    'Dermatology': [
        'SkinHeal',
        'DermaCare',
        'SkinPlus',
        'DermaSafe',
        'SkinGlow',
        'DermaLife',
        'SkinPrime',
        'DermaX',
        'GlowCare',
        'SkinShield'
    ]
}

products = []

product_id = 1

for category, names in categories.items():

    for product in names:

        products.append([
            product_id,
            product,
            category,
            round(random.uniform(50, 500), 2)
        ])

        product_id += 1

products_df = pd.DataFrame(
    products,
    columns=[
        'product_id',
        'product_name',
        'category',
        'price'
    ]
)

products_df.head()
```

```python
products_df.to_csv(
    '../01_Dataset/raw_data/products.csv',
    index=False
)

print("Products CSV Saved Successfully")
```

```python
regions = [
    'Hyderabad',
    'Bangalore',
    'Chennai',
    'Mumbai',
    'Pune',
    'Delhi'
]

medical_reps = []

for i in range(1, 101):

    joining_date = fake.date_between(
        start_date='-10y',
        end_date='today'
    )

    medical_reps.append([
        i,
        fake.name(),
        random.choice(regions),
        random.randint(1, 15),
        joining_date
    ])

medical_reps_df = pd.DataFrame(
    medical_reps,
    columns=[
        'mr_id',
        'mr_name',
        'region',
        'experience_years',
        'joining_date'
    ]
)

medical_reps_df.head()
```

```python
medical_reps_df.shape
```

```python
# Territory performance multipliers

territory_multiplier = {
    'Hyderabad': 1.30,
    'Bangalore': 1.20,
    'Mumbai': 1.15,
    'Delhi': 1.10,
    'Chennai': 1.00,
    'Pune': 0.90
}
```

```python
# Product Price Lookup

product_price = dict(
    zip(
        products_df['product_id'],
        products_df['price']
    )
)

product_category = dict(
    zip(
        products_df['product_id'],
        products_df['category']
    )
)

doctor_specialization = dict(
    zip(
        doctors_df['doctor_id'],
        doctors_df['specialization']
    )
)

mr_experience = dict(
    zip(
        medical_reps_df['mr_id'],
        medical_reps_df['experience_years']
    )
)
```

```python
sales = []

for sale_id in range(1, 50001):

    mr_id = random.randint(1, 100)

    doctor_id = random.randint(1, 500)

    product_id = random.randint(1, 50)

    sale_date = fake.date_between(
        start_date='-2y',
        end_date='today'
    )

    territory = random.choice([
        'Hyderabad',
        'Bangalore',
        'Mumbai',
        'Delhi',
        'Chennai',
        'Pune'
    ])

    experience_factor = (
        mr_experience[mr_id] / 10
    )

    territory_factor = (
        territory_multiplier[territory]
    )

    units_sold = max(
        1,
        int(
            random.randint(5, 50)
            * experience_factor
            * territory_factor
        )
    )

    revenue = round(
        units_sold *
        product_price[product_id],
        2
    )

    sales.append([
        sale_id,
        sale_date,
        mr_id,
        doctor_id,
        product_id,
        units_sold,
        revenue,
        territory
    ])

sales_df = pd.DataFrame(
    sales,
    columns=[
        'sales_id',
        'sale_date',
        'mr_id',
        'doctor_id',
        'product_id',
        'units_sold',
        'revenue',
        'territory'
    ]
)

sales_df.head()
```

```python
sales_df.to_csv(
    '../01_Dataset/raw_data/sales.csv',
    index=False
)

print("Sales CSV Saved Successfully")
```
